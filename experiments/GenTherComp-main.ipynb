### Establish src directory

```python
from pathlib import Path
import importlib
import shutil
import subprocess
import sys

REPO_URL = "https://github.com/ar3ene/research.git"
REPO_DIR = Path("/content/research")
SPARSE_PATH = "src/GenTherComp"


def _ensure_package(package: str, module_name: str | None = None) -> None:
    module = module_name or package
    try:
        importlib.import_module(module)
    except ModuleNotFoundError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", package])


def _run(cmd: list[str]) -> None:
    print("$", " ".join(cmd))
    subprocess.check_call(cmd)


_ensure_package("numpy")

# 在每次 Colab 新实例里只同步 src/GenTherComp（不依赖 /content/research）
if (REPO_DIR / ".git").exists():
    print(f"Updating sparse repo: {REPO_DIR}")
    _run(["git", "-C", str(REPO_DIR), "remote", "set-url", "origin", REPO_URL])
    _run(["git", "-C", str(REPO_DIR), "sparse-checkout", "init", "--cone"])
    _run(["git", "-C", str(REPO_DIR), "sparse-checkout", "set", SPARSE_PATH])
    _run(["git", "-C", str(REPO_DIR), "pull", "--ff-only", "origin", "main"])
else:
    if REPO_DIR.exists():
        print(f"Repo dir exists but is not a git repo, reinitializing: {REPO_DIR}")
        shutil.rmtree(REPO_DIR)
    print(f"Cloning sparse repo into: {REPO_DIR}")
    _run([
        "git",
        "clone",
        "--depth",
        "1",
        "--filter=blob:none",
        "--sparse",
        REPO_URL,
        str(REPO_DIR),
    ])
    _run(["git", "-C", str(REPO_DIR), "sparse-checkout", "set", SPARSE_PATH])

src_pkg = REPO_DIR / "src" / "GenTherComp"
if not src_pkg.exists():
    raise FileNotFoundError(f"未找到 {src_pkg}，请检查仓库路径是否正确。")

if str(REPO_DIR) not in sys.path:
    sys.path.insert(0, str(REPO_DIR))

print(f"Using repo_root: {REPO_DIR}")
print(f"Source package: {src_pkg}")
print(f"Python executable: {sys.executable}")
```
